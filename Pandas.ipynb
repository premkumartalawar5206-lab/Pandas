### Handling Missing Values

```python
df_na=pd.DataFrame({'a':[1,np.nan,3],'b':[np.nan,2,3]})
```

```python
df_na
```

```python
df_na.isna() #Boolean mask of missing values
```

```python
df_na.notna() #revers of .isna()
```

```python
df_na.isnull().sum() #counts missing values by cloumns
```

```python
df_na.dropna()
```

```python
df_na.dropna(axis=1) #use to drop columns
```

```python
df_na.fillna(0.5) #fill null values
```

```python
df_na.ffill() #forward fill
```

```python
df_na.bfill() #backward fill
```
